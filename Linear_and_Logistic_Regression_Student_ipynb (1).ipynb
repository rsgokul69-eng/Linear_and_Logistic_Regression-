```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression as SkLinearRegression
from sklearn.linear_model import LogisticRegression as SkLogisticRegression
```

```python
class LinearRegression:


  def fit(self, X, y):

    x_mean = np.mean(X)
    y_mean = np.mean(y)

    numerator = np.sum((X - x_mean) * (y - y_mean))
    denominator = np.sum((X - x_mean) ** 2)

    self.m = numerator / denominator
    self.c = y_mean - (self.m * x_mean)

  def predict(self, x):
    y_pred = (self.m * x) + self.c
    return y_pred
```

```python
class LogisticRegression:

  def fit(self, X, y):
    self.m = 0
    self.c = 0

    learning_rate = 0.01
    epochs = 1000
    n = len(X)

    for i in range(epochs):
      y_pred = self.m * X + self.c
      probability = 1 / (1 + np.exp(-y_pred))

      error = probability - y

      dm = np.sum(error * X) / n
      dc = np.sum(error) / n

      self.m = self.m - learning_rate * dm
      self.c = self.c - learning_rate * dc

  def predict(self, x):
    y_pred = (self.m * x) + self.c
    probability = 1 / (1 + np.exp(-y_pred))
    threshold = 0.5
    prediction = 1 if probability >= threshold else 0
    return (probability, prediction)
```

```python
df = pd.DataFrame(
    {
        "study_hours" : [1, 2, 3, 4, 5, 6, 7, 8, 9],
        "exam_score" : [35, 40, 45, 50, 55, 65, 70, 80, 90]
    }
)


X = df["study_hours"]
y = df["exam_score"]

linear_reg = LinearRegression()
linear_reg.fit(X, y)

x = 5
prediction = linear_reg.predict(x)

print(f"Study Hours: {x}\nPredicted Exam Score: {prediction}")

sk_linear_model = SkLinearRegression()

sk_linear_model.fit(df[["study_hours"]], y)

sk_prediction = sk_linear_model.predict(
    pd.DataFrame({"study_hours": [5]})
)
print("-------------")
print("Sklearn Linear Regression")
print("Prediction:", sk_prediction[0])

print("-----------------------------------------------------------------")

y_class = (df["exam_score"] >= 60).astype(int)

logistic_reg = LogisticRegression()

logistic_reg.fit(X, y_class)
x = 4
prob, log_pred = logistic_reg.predict(x)

print(f"Study Hours: {x}\nPrediction class: {log_pred}")

if log_pred == 1:
  print("Pass")
else:
  print("Fail")

print("-------------")
sk_logistic_model = SkLogisticRegression()

sk_logistic_model.fit(df[["study_hours"]], y_class)

test = pd.DataFrame({"study_hours": [4]})

sk_probability = sk_logistic_model.predict_proba(test)[0][1]

sk_class = sk_logistic_model.predict(test)[0]

print("Sklearn Logistic Regression")
print("Predicted class:", sk_class)
```
